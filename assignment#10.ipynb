```python
result = dataset[(dataset["Sales"] > 500) & (dataset["Quantity"] > 5)]

print(result)
```

```python
result = dataset[(dataset["Discount"] > 0.5) | (dataset["Profit"] < 0)]

print(result)
```

```python
result = dataset.loc[dataset["Region"] == "East", ["Ship Mode", "Profit"]]

print(result)
```

```python
dataset["Profit Margin"] = dataset["Profit"] / dataset["Sales"]

result = dataset[dataset["Profit Margin"] > 0.3]

print(result)
```

```python
result = dataset[(dataset["Category"] == "Furniture") & (dataset["Sub-Category"] != "Chairs")]

print(result)
```
