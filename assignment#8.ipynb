```python
class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describeEmployee(self):
        print("===================================")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Years of Experience:", self.year_of_experience)
        print("===================================\n")

    def updateYearOfExperience(self, new_experience):
        self.year_of_experience = new_experience

    def increaseSalary(self, amount):
        self.salary += amount

emp1 = Employee("Ali", 101, "Lahore", 50000, "Manager", 2018, "MBA", 8)
emp2 = Employee("Ahmed", 102, "Karachi", 45000, "Accountant", 2019, "B.Com", 6)
emp3 = Employee("Sara", 103, "Islamabad", 60000, "Engineer", 2017, "BS Engineering", 9)
emp4 = Employee("Ayesha", 104, "Multan", 40000, "Teacher", 2020, "M.Ed", 5)
emp5 = Employee("Usman", 105, "Faisalabad", 55000, "Developer", 2016, "BSCS", 10)
emp6 = Employee("Hamza", 106, "Peshawar", 48000, "Designer", 2021, "BFA", 4)
emp7 = Employee("Fatima", 107, "Quetta", 52000, "HR Officer", 2018, "MBA HR", 7)
emp8 = Employee("Zain", 108, "Sialkot", 47000, "Sales Officer", 2022, "BBA", 3)
emp9 = Employee("Noor", 109, "Rawalpindi", 65000, "Doctor", 2015, "MBBS", 11)
emp10 = Employee("Bilal", 110, "Gujranwala", 58000, "Supervisor", 2017, "BS", 8)

employees = [emp1, emp2, emp3, emp4, emp5,
             emp6, emp7, emp8, emp9, emp10]

print("Employee Details")
for emp in employees:
    emp.describeEmployee()

emp1.updateYearOfExperience(9)
emp1.increaseSalary(5000)

emp5.updateYearOfExperience(11)
emp5.increaseSalary(7000)

print("After Updates\n")

emp1.describeEmployee()
emp5.describeEmployee()
```
