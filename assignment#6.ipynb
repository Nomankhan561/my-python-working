```python
secret_word = "python"

guess = ""

while guess != secret_word:
    guess = input("Guess the word: ")

print("🎉 You guessed it!")
```

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]

count = 0

for user in users:
    print(f"Email sent to {user}")
    count += 1

    if count == 3:
        break
```

```python
temps = [22, 25, 30, 45, 28]
critical = 40

for temp in temps:
    print(f"Temperature: {temp}°C")

    if temp >= critical:
        print("⚠️ Critical temperature reached! Stopping monitoring.")
        break
```

```python
cart = [100, 250, 50, 300]

total = 0

for price in cart:
    total += price

print("Total Price:", total)
```

```python
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if not customer["paid"]:
        continue

    print(f"Serving {customer['name']}")
```

```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

for email in emails:
    domain = email.split("@")[1]
    print(domain)
```

```python
tasks = [
    {"task": "Pay bills", "status": "done"},
    {"task": "Submit report", "status": "pending"},
    {"task": "Clean house", "status": "pending"}
]

for task in tasks:
    if task["status"] == "pending":
        print(task["task"])
```
