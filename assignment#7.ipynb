```python
def add_numbers(num1, num2):
    return num1 + num2
result = add_numbers(10, 15)
print("The sum is:", result)
```

```python
def currency_converter(amount, exchange_rate=280):
    converted_amount = amount * exchange_rate
    return f"{amount} USD is equivalent to {converted_amount} PKR"
print(currency_converter(100))
```

```python
def calculate_price(price, quantity):
    return price * quantity
total = calculate_price(50, 4)
print("Total Price:", total)
```

```python
def calculate_rectangle(length, width):

    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    return calculate_area(), calculate_perimeter()

area, perimeter = calculate_rectangle(5, 3)

print("Area:", area)
print("Perimeter:", perimeter)
```

```python
balance = 1000

try:
    amount = float(input("Enter amount to withdraw: "))

    if amount <= 0:
        print("Amount must be more than zero.")
    elif amount > balance:
        print("Insufficient balance.")
    else:
        balance -= amount
        print("Withdrawal successful.")
        print("Remaining balance:", balance)

except ValueError:
    print("Please enter a valid number.")

finally:
    print("Thank you for using the ATM.")
```

```python
available_seats = 5

try:
    tickets = int(input("How many tickets do you want to book? "))

    if tickets <= 0:
        print("You must book at least 1 ticket.")
    elif tickets > available_seats:
        print("Not enough seats available.")
    else:
        print(f"{tickets} ticket(s) booked successfully.")

except ValueError:
    print("Please enter a valid number.")

finally:
    print("Thank you for booking with us.")
```

```python
class Light:

    def turn_on(self):
        print("The light is now ON.")

    def check_status(self):
        print("The light is bright and working.")


# Create an object of the Light class
my_light = Light()

# Call both methods
my_light.turn_on()
my_light.check_status()
```

```python
class Student:

    def __init__(self, name, grade, age):
        self.name = name
        self.grade = grade
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}, I am {self.age} years old.")

    def show_grade(self):
        print(f"My grade is {self.grade}.")
        
student1 = Student("Alice", "A", 16)

# Call both methods
student1.introduce()
student1.show_grade()
```
